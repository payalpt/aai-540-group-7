## Import PyAthena

```python
from pyathena import connect
```

## Ceate Athena Database

```python
database_name = "audio_emotions"
```

```python
# Set S3 staging directory -- this is a temporary directory used for Athena queries
s3_staging_dir = "s3://sagemaker-us-east-1-218117716191/athena/staging/" #Our directory
```

```python
conn = connect(s3_staging_dir=s3_staging_dir, region_name=region, )
```

```python
statement = "CREATE DATABASE IF NOT EXISTS {}".format(database_name)
print(statement)
```

```python
import pandas as pd

pd.read_sql(statement, conn)
```

## Verify The Database Has Been Created Successfully

```python
statement = "SHOW DATABASES"

df_show = pd.read_sql(statement, conn)
df_show.head(5)
```

```python
if database_name in df_show.values:
    ingest_create_athena_db_passed = True
```
